Business Problem Statement

Analyze Airbnb listing data to understand pricing patterns, room-type and location differences, availability, and guest review activity, in order to identify key factors influencing listing performance and uncover opportunities for hosts to improve pricing and competitiveness.

Key Business Questions

What factors are associated with Airbnb listing prices?

# Step:2 Data Cleaning

Data Cleaning Tasks

# Checking Missing Values

```python
df.isnull().sum()
```

```python
df.info()
```

# Handling Missing values

```python
df['last review'] = pd.to_datetime(df['last review'], errors='coerce')
```

```python
df['reviews per month'] = df['reviews per month'].fillna(0)
df['last review_missing'] = df['last review'].isna()
```

```python
df.drop(columns=['license','house_rules'], errors='ignore',inplace=True)
```

```python
df.dropna(subset=['NAME', 'host name', 'host_identity_verified',
       'neighbourhood group', 'neighbourhood', 'lat', 'long', 'country',
       'country code', 'instant_bookable', 'cancellation_policy',
       'Construction year', 'price', 'service fee', 'minimum nights',
       'calculated host listings count',
       'availability 365'], inplace=True)
```

```python
# Removing dolor sign and converting price to float
df['price'] = df['price'].replace('[\$,]', '', regex=True).astype(float)
df['service fee'] = df['service fee'].replace('[\$,]', '', regex=True).astype(float)
```

```python
df.isnull().sum()
```

```python

df.loc[df['availability 365'] > 365, 'availability 365'] = np.nan
df_availability = df.dropna(subset=['availability 365']).copy()
```

```python
df.drop_duplicates(inplace=True)
```

# Step:- 4 Visualization

# What is the distribution of listing prices?

```python
plt.figure(figsize=(10,6))
sns.histplot(df['price'],bins=50,kde=True,color='green')
plt.title('Distribution of listing Price')
plt.xlabel('Price $')
plt.ylabel('Frequency')
plt.show()
```

Most Airbnb listings are concentrated around the mid-price range.
The median price is around $624, showing the typical listing price.
Prices vary considerably across listings.
This helps identify the common market price range and potential premium listings

# How are different Room types distributed?

```python
df['room type']
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x="room type",data = df, color="purple")
plt.title('Room Type Distribution')
plt.xlabel('Room Type')
plt.ylabel('Count')
plt.show()
```

The chart shows which room types have the largest number of listings.

# How are listing distributed across different nighborhoods?

```python
plt.figure(figsize=(8, 8))

sns.countplot(
    y="neighbourhood group",
    data=df,color='hotpink',
    order=df['neighbourhood group'].value_counts().index
)

plt.title('Neighbourhood Group Distribution')
plt.xlabel('Number of Listings')
plt.ylabel('Neighbourhood Group')

plt.show()
```

The chart compares Airbnb listing counts across neighbourhood groups.
Some areas have substantially more listings than others.
This highlights locations with higher Airbnb supply and competition.
Business question: Which areas have the highest concentration of listings?

# What is the relationship between Price and room type?


```python
plt.figure(figsize=(10,6))

sns.boxplot(
    x='room type',
    y='price',
    hue='room type',
    data=df,
    palette='Set1',
    legend=False
)

plt.title('Price vs Room Type')
plt.xlabel('Room Type')
plt.ylabel('Price ($)')

plt.show()
```

Different room types show different pricing patterns.
The chart compares their median prices and overall price variation.
This helps identify which accommodation types can command higher prices.
Business question: Does room type influence Airbnb pricing?

# How has the number of reviews change over with time

```python
df['last review']= pd. to_datetime(df['last review'])
reviews_over_time = df.groupby(df['last review'].dt.to_period('M')).size()
plt.figure(figsize=(12, 6))
reviews_over_time.plot(kind='line', color='red')
plt.title('Number of Listings by Last Review Month')
plt.xlabel('Time')
plt.ylabel('Number of reviews')
plt.show()
```

The chart shows how Airbnb listings are distributed according to their last review month.
It highlights periods with higher or lower review activity.
This helps understand the timing of guest engagement.
Business question: When were Airbnb listings most recently reviewed?

# Which areas have high demand but low availability?

```python
df.groupby('neighbourhood group').agg({'number of reviews': 'mean','availability 365': 'mean'
                                       }).sort_values('number of reviews', ascending=False)
```

The comparison shows average reviews and availability across neighbourhood groups.
Areas with more reviews may indicate stronger guest activity.
Comparing this with availability can highlight potentially high-demand areas.
Business question: Which locations may have high demand but limited availability?

# KPIs

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(14, 4))

kpis = {
    'Total Listings': f"{len(df):,}",
    'Average Price': f"${df['price'].mean():,.2f}",
    'Median Price': f"${df['price'].median():,.2f}",
    'Total Reviews': f"{df['number of reviews'].sum():,}"
}

# Colors for each KPI box
colors = ['#4C78A8', '#F58518', '#54A24B', '#E45756']

ax.axis('off')

for i, ((title, value), color) in enumerate(zip(kpis.items(), colors)):
    
    ax.text(
        0.125 + i * 0.25,
        0.5,
        f"{title}\n\n{value}",
        ha='center',
        va='center',
        fontsize=15,
        fontweight='bold',
        color='white',
        bbox=dict(
            boxstyle='round,pad=0.8',
            facecolor=color,
            edgecolor='black',
            linewidth=1.5
        )
    )

plt.tight_layout()
plt.show()
```

The dataset contains 98,840 listings after cleaning.
Average listing price is approximately $625, while the median is approximately $624.
These KPIs provide a quick overview of marketplace size and pricing.
Business question: What is the overall size and pricing level of the Airbnb marketplace?
